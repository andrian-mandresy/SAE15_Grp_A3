```python
import pandas as pd
import folium
from folium.plugins import FastMarkerCluster

print("Instantly loading the complete CSV file...")
# Load the pre-processed CSV file containing all merged data and coordinates
ssaenpt = pd.read_csv('antenne_avec_coords.csv', sep=';', low_memory=False)

print("Quickly generating the HTML structure for popups with all details...")
# Combine columns into a single HTML string for each row to display inside the map popups
ssaenpt["popup_html"] = (
    '<div style="font-family: Arial; font-size: 12px; width: 220px;">'
    + "<b>Lieu :</b> " + ssaenpt['ADR_LB_LIEU'].astype(str) + "<br>"
    + "<b>Date de service :</b> " + ssaenpt['DTE_EN_SERVICE'].astype(str) + "<br>"
    + "<b>Hauteur :</b> " + ssaenpt['SUP_NM_HAUT'].astype(str) + " m<br>"
    + "<b>Code postal :</b> " + ssaenpt['ADR_NM_CP'].astype(str) + "<br>"
    + "<b>Propriétaire :</b> " + ssaenpt['TPO_LB'].astype(str) + "<br>"
    + "<b>Type de support :</b> " + ssaenpt['NAT_LB_NOM'].astype(str) + "<br>"
    + "<b>Type d'antenne :</b> " + ssaenpt['TAE_LB'].astype(str) + "<br>"
    + "<b>Exploitant :</b> " + ssaenpt['ADM_LB_NOM'].astype(str)
    + "</div>"
)

# Extract latitude, longitude, and popup HTML into a simple list for FastMarkerCluster
data_list = ssaenpt[['latitude', 'longitude', 'popup_html']].values.tolist()

print("Creating the map...")
# Initialize the base map centered on France with Esri Satellite imagery
carte_satellite = folium.Map(
    location=[46.603354, 1.888334], 
    zoom_start=6,
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}',
    attr='Esri'
)

# JavaScript callback function to render markers and popups quickly in the browser
callback = """
function (row) {
    var marker = L.marker(new L.LatLng(row[0], row[1]));
    var icon = L.AwesomeMarkers.icon({icon: 'signal', prefix: 'fa', markerColor: 'blue'});
    marker.setIcon(icon);
    var popup = L.popup({maxWidth: 250}).setContent(row[2]);
    marker.bindPopup(popup);
    return marker;
};
"""

print("Injecting all points into the map cluster...")
# Add all points to the map using FastMarkerCluster for high performance
FastMarkerCluster(data=data_list, callback=callback).add_to(carte_satellite)

# Save the final interactive map to an HTML file
output_file = "carte_antennes_integral.html"
carte_satellite.save(output_file)
print(f"🎉 Done! Complete map successfully generated in: {output_file}")
```
